```python
import requests
import pandas as pd
import time
```

```python
all_records = []

limit = 10
offset = 0
total = 3402

while offset < total:

    api_url = (
        "https://api.data.gov.in/resource/"
        "3b01bcb8-0b14-4abf-b6f2-c1bfd384ba69"
        f"?api-key={API_KEY}"
        "&format=json"
        f"&offset={offset}"
        f"&limit={limit}"
    )

    try:
        response = requests.get(
            api_url,
            headers={"User-Agent": "Mozilla/5.0"},
            timeout=180
        )

        if response.status_code == 200:

            data = response.json()
            records = data.get("records", [])

            all_records.extend(records)

            print(
                f"Fetched {len(records)} | "
                f"Total collected: {len(all_records)} / {total}"
            )

            if len(records) == 0:
                print("No more records available.")
                break

            offset += len(records)

            time.sleep(1)

        else:
            print("Error:", response.status_code)
            break

    except requests.exceptions.Timeout:
        print(f"Timeout at offset {offset}. Retrying...")
        time.sleep(5)
        continue

print("Finished!")
print("Total records collected:", len(all_records))
```

```python
df = pd.DataFrame(all_records)

print("Shape:", df.shape)
```
